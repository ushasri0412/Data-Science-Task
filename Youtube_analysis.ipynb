```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("https://github.com/Taj1920/Data_Analysis/raw/refs/heads/main/Youtube_analysis/YOUTUBE%20CHANNELS%20DATASET.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Ranking"].unique()
```

```python
#Handling null values
df["Ranking"]=df["Ranking"].fillna("101")
df["Username"]=df["Username"].fillna("Unknown")

```

```python
df["Subscribers"].unique()
```

```python
df["Subscribers"]=df["Subscribers"].str.strip("M")
df["Subscribers"]=df["Subscribers"].astype(float)
df["Subscribers"].unique()
```

```python
df["Subscribers"]=df["Subscribers"].fillna(df["Subscribers"].mean())
df["Subscribers"].unique()
```

```python
df["Uploads"].unique()
```

```python
df["Uploads"]=df["Uploads"].fillna("0")
df["Uploads"].unique()
```

```python
df["Views"].unique()
```

```python
df["Views"]=df["Views"].fillna("0")
df["Views"].unique()
```

```python
df["Country"].unique()
```

```python
df["Country"]=df["Country"].fillna("unknown_country")
df["Country"].unique()
```

```python
df.isnull().sum()
```

```python
df.shape
df.drop_duplicates(inplace=True)
df.shape
```

```python
df["Views"]=df["Views"].str.replace(",","")
```

```python
df["Uploads"]=df["Uploads"].str.replace(",","")
```

```python
df
```

```python
# df.drop(["Country","Username"],axis=1)
```

```python
plt.plot(df["Ranking"],df["Subscribers"])
plt.xlabel("Ranking")
plt.ylabel("Subscribers")
plt.title("Ranking vs Subscribers (Linear Graph)")

plt.show()
```

```python
plt.bar(df["Country"],df["Subscribers"])
plt.xlabel("Country")
plt.ylabel("Subscribers")
plt.title("Country vs Subscribers (Bar Graph)")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.bar(df["Country"],df["Username"])
plt.xlabel("Country")
plt.ylabel("Username")
plt.title("Country vs Username (Bar Graph)")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Subscribers"], df["Views"])

plt.xlabel("Subscribers")
plt.ylabel("Views")
plt.title("Subscribers vs Views")

plt.show()
```

```python
top5 = df.sort_values("Subscribers", ascending=False).head(5)
plt.figure(figsize=(6,6))
plt.pie(top5["Subscribers"],
        labels=top5["Ranking"],
        autopct='%1.1f%%')
plt.title("Top 5 Subscribers Share")
plt.show()
```
