```python
#import necessary Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers=pd.read_csv("/content/customers.csv")
orders=pd.read_csv("/content/orders.csv")
products=pd.read_csv("/content/products.csv")
```

```python
df=pd.merge(customers,orders,on="customer_id",how="inner")
df=pd.merge(df,products,on="product_id",how="inner")
df
```

```python
#EDA
df.info()
```

```python
#converting date columns which are in object form to their original date time format
df["signup_date"]=pd.to_datetime(df["signup_date"])
df["order_date"]=pd.to_datetime(df["order_date"])
df.info()
```

```python
#we should create a new column for revenue
#formula : revenue =price * quantity
df["revenue"] = df["price"] * df["quantity"]
df.head()
```

```python
#Revenue trend by order date
revenue=df.groupby("order_date")["revenue"].sum() # this will give sum sorting -->> .sort_values
revenue=pd.DataFrame(revenue).reset_index()
revenue
plt.plot(revenue["order_date"],revenue["revenue"],marker="o",mfc="orange",mec="Black")
plt.xlabel("Order Date")
plt.ylabel("Revenue")
plt.title("Revenue Trend by order date")
plt.show()
```

#Sales are dropping in mid  may june and july

```python
df["city"].unique()
```

```python
#Revenue trend by city
revenue=df.groupby("city")["revenue"].sum() # this will give sum sorting -->> .sort_values
revenue=pd.DataFrame(revenue).reset_index()
revenue
plt.plot(revenue["city"],revenue["revenue"],marker="o",mfc="green",mec="Black")
plt.xlabel("city")
plt.ylabel("Revenue")
plt.title("Revenue Trend by cities")
plt.show()

```

#Revenue is dropping in Bangalore compared to other cities

```python
#Revenue trend by category
revenue=df.groupby("category")["revenue"].sum() # this will give sum sorting -->> .sort_values
revenue=pd.DataFrame(revenue).reset_index()
revenue
plt.plot(revenue["category"],revenue["revenue"],marker="o",mfc="green",mec="Black")
plt.xlabel("category")
plt.ylabel("Revenue")
plt.title("Revenue Trend by Category")
plt.show()

```

#Revenue is dropping in Fashion Category.

```python
#count customers based on order id ( count of orders per customer)
df["customer_id"].count()
df.groupby("customer_id")["order_id"].count()
```

# Most of the customers are ordering less

```python
#Revenue trend by product_name
revenue=df.groupby("product_name")["revenue"].sum() # this will give sum sorting -->> .sort_values
revenue=pd.DataFrame(revenue).reset_index()
revenue
plt.plot(revenue["product_name"],revenue["revenue"],marker="o",mfc="green",mec="Black")
plt.xlabel("product Name")
plt.ylabel("Revenue")
plt.title("Revenue Trend by product name")
plt.show()
```

# revenue is very less for shoes and watch

# Insights :
1. Revenue got dropped in last 2 months - june and july

2. Banglore Contributed more in revenue drop

3. Fashion Category recorded less revenue  

4. Repeated customers ordered less

5. shoe and watch declined slowly
