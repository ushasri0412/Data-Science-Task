```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Cleaning/USA_presidents.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.shape
df.drop_duplicates(inplace=True)
df.shape
```

```python
# df=df.drop("Unnamed: 0",axis=1)
# df
# df=df.drop("Unnamed: 0.1",axis=1)
# df
df=df.drop("S.No.",axis=1)
df
```

```python
df["president"].unique()
```

```python
df["prior"].unique()
```

```python
d={"president":"President Name","prior": "Priotity_position",	"party": "Party Type"	,"vice": "Vice President","salary" :"Salary","date updated": "Date Updated","date created" :"Date Created"}
df.rename(columns=d,inplace=True)
df
```

```python
df["Priotity_position"] = df["Priotity_position"].str.replace("â€“", "-")
df["Priotity_position"].unique()
```

```python
df["Party Type"].unique()
```

```python
df["Party Type"]=df["Party Type"].str.strip("   April 4, 1841  â€“  September 13, 1841")
df["Party Type"].unique()
```

```python
df["Party Type"] = df["Party Type"].str.replace("Republicans", "Republican")
df["Party Type"].unique()
```

```python
df["Party Type"] = df["Party Type"].str.replace("Demorcatic", "Democratic")
df["Party Type"].unique()
```

```python
df["Party Type"] = df["Party Type"].str.replace("Democratic-  Republican", "Democratic-Republican")
df["Party Type"].unique()
```

```python
df["Vice President"].unique()
```

```python
df["Vice President"]=df["Vice President"].str.strip("     ")
df["Vice President"].unique()
```

```python
df["Vice President"]=df["Vice President"].str.replace("John C.     Calhoun","John C.Calhoun")
df["Vice President"].unique()
```

```python
df["Vice President"]=df["Vice President"].str.replace("John C. Calhoun","John C.Calhoun")
df["Vice President"].unique()
```

```python
df["Vice President"]=df["Vice President"].str.replace("George         M. Dallas","George M. Dallas")
df["Vice President"].unique()
```

```python
df["Salary"].unique()
```

```python
df["Salary"]=df["Salary"].str.strip("$")
df["Salary"].unique()
```

```python
df["Salary"]=df["Salary"].str.replace(",","")
df["Salary"].unique()
```

```python
df["Salary"]=df["Salary"].astype(float)
df["Salary"]=df["Salary"].astype(int)
df["Salary"].unique()
```
